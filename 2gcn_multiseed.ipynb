```python
edges = re.findall(r'^(\d+),(\d+),-1$', content, flags=re.MULTILINE)
```

```python
raw_nodes = re.findall(r'^(\d+),"([^"]+)",(?:"?([^",\s]+)"?),(\d+)$', content, flags=re.MULTILINE)
```

```python
raw_nodes
```

```python
node_properties = []
```

```python
for item in raw_nodes:
    node_id = item[0]
    desc = item[1]
    node_type = item[2]
    val = item[3]

    if node_type == 'OR':
        shape = 'diamond'
    elif node_type == 'AND':
        shape = 'ellipse'
    else:
        shape = 'box'

    formatted_string = f'label="{node_id}:{desc}:{val}",shape={shape}'
    node_properties.append(formatted_string)
```

```python
node_properties
```

```python
len(node_properties)
```

```python
node_dict={}
nodes = [item[0] for item in raw_nodes]
```

```python
nodes
```

```python
for item in node_properties:
    label_match=re.search(r'label="(.*)"',item)
    label_text=label_match.group(1)

    parts=label_text.split(':')
    node_id=nodes.index(parts[0])
    statement=parts[1]

    match=re.match(r'(.+)\((.*)\)',statement)
    predicate=match.group(1).strip()
    attr_text=match.group(2).strip()

    if ',' in attr_text:
        attributes=attr_text.split(',')
    else:
        attributes=attr_text.split()

    clean_attributes=[]
    for a in attributes:
        clean_attributes.append(a.strip("'\""))
    attributes=clean_attributes

    shape_match=re.search(r'shape=(\w+)',item)
    shape=shape_match.group(1)

    node_dict[node_id]={
        'predicate':predicate,
        'attributes':attributes,
        'shape':shape
    }
```

### GCN Model — Multi-Seed Training

```python
in_dim = X_train.shape[-1]
hidden_dim = 20
out_dim = 1

num_epochs = 150
seeds = [42, 123, 999, 7, 2024]
```

```python
def train_and_evaluate_gcn(seed):

    torch.manual_seed(seed)

    model = GCN(in_dim, hidden_dim, out_dim).to(device)

    edges = edge_index_t.to(device)
    action_mask = torch.tensor(action_node_idx, dtype=torch.long, device=device)

    X_train_dev = X_train.to(device)
    Y_train_dev = Y_train.to(device)
    X_val_dev = X_val.to(device)
    Y_val_dev = Y_val.to(device)

    num_class_0 = (Y_train_dev == 0).sum().item()
    num_class_1 = (Y_train_dev == 1).sum().item()
    pos_weight = torch.tensor([num_class_0 / num_class_1], dtype=torch.float32).to(device)
    criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

    optimizer = optim.Adam(model.parameters(), lr=0.001)

    def evaluate(X_data, Y_data):
        model.eval()
        with torch.no_grad():
            output = model(X_data, edges)
            loss = criterion(output[:, action_mask], Y_data)
            pred = (torch.sigmoid(output[:, action_mask]) > 0.5).float()
            accuracy = (pred == Y_data).sum().item() / Y_data.numel()
        return loss.item(), accuracy

    best_val_loss = float('inf')
    best_state = None

    train_dataset = TensorDataset(X_train_dev, Y_train_dev)
    train_loader = DataLoader(train_dataset, batch_size=30, shuffle=False)

    for epoch in range(num_epochs):
        model.train()

        for batch_X, batch_Y in train_loader:
            output = model(batch_X, edges)
            loss = criterion(output[:, action_mask], batch_Y)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

        val_loss, val_acc = evaluate(X_val_dev, Y_val_dev)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = {k: v.clone() for k, v in model.state_dict().items()}

    model.load_state_dict(best_state)
    model.eval()

    torch.save(model.state_dict(), f"best_gcn_fixed_seed{seed}.pt")

    all_preds, all_probs, all_labels = [], [], []

    with torch.no_grad():
        for i in range(len(X_test)):
            x_i = X_test[i].to(device)
            y_i = Y_test[i].to(device)

            output = model(x_i, edges)
            probs = torch.sigmoid(output[action_mask])
            preds = (probs > 0.5).float()

            all_probs.append(probs.cpu())
            all_preds.append(preds.cpu())
            all_labels.append(y_i.cpu())

    all_probs = torch.stack(all_probs).flatten().numpy()
    all_preds = torch.stack(all_preds).flatten().numpy()
    all_labels = torch.stack(all_labels).flatten().numpy()

    conf_matrix = confusion_matrix(all_labels, all_preds)
    TN, FP, FN, TP = conf_matrix.ravel()
    precision, recall, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average='binary', zero_division=0)
    fpr, tpr, _ = roc_curve(all_labels, all_probs)
    roc_auc = auc(fpr, tpr)
    accuracy = (all_preds == all_labels).sum() / len(all_labels)

    return {
        'seed': seed,
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'auc': roc_auc,
        'fpr': FP / (FP + TN),
        'fnr': FN / (FN + TP),
        'precision_macro': _prec(all_labels, all_preds, average='macro', zero_division=0),
        'recall_macro': _rec(all_labels, all_preds, average='macro', zero_division=0),
        'f1_macro': _f1(all_labels, all_preds, average='macro', zero_division=0),
    }
```

```python
print("\nDone. Final used_indices per keyword:", used_indices)
print(y_malic.sum())
```

```python
results = []

for seed in seeds:
    print(f"--- Training with seed {seed} ---")
    metrics = train_and_evaluate_gcn(seed)
    results.append(metrics)
    print(metrics)
    print()
```
